# Phase 1 — EduRABSA Dataset Inspection

## Objective
Inspect the actual EduRABSA dataset before making any assumptions about schema, labels, annotations, splits, or capabilities.

**Rule:** This notebook does not clean, transform, rename, delete, or model the raw data.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
DATA_ROOT = PROJECT_ROOT / 'data' / 'raw' / 'edurabsa'

print('Project root:', PROJECT_ROOT)
print('EduRABSA root:', DATA_ROOT)
print('Exists:', DATA_ROOT.exists())
```

## 1. Recursively discover files


```python
files = sorted(p for p in DATA_ROOT.rglob('*') if p.is_file())
print(f'Total files: {len(files)}')

for path in files:
    print(path.relative_to(PROJECT_ROOT))
```

## 2. File type and size inventory


```python
from collections import Counter

extensions = Counter(p.suffix.lower() or '[no extension]' for p in files)
for ext, count in sorted(extensions.items()):
    print(f'{ext}: {count}')
```

## 3. Inspect supported data files

The next step is intentionally conservative. We inspect structure and quality but do not alter the source data.


```python
# Run the reusable inspector from the project scripts directory.
# From the project root, the same inspection can also be run with:
# python scripts/inspect_edurabsa.py

import subprocess, sys
result = subprocess.run(
    [sys.executable, 'scripts/inspect_edurabsa.py'],
    capture_output=True,
    text=True,
)
print(result.stdout)
if result.stderr:
    print('STDERR:\n', result.stderr)
```
